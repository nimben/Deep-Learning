P(Rain) on z cloudy day

```python
P_cloudy = 0.40
P_rain = 0.20
P_cloudy_given_rain = 0.85


P_rain_given_cloudy = (
    P_cloudy_given_rain * P_rain
) / P_cloudy

print("Probability of rain given cloudy:",
      P_rain_given_cloudy)

print("Percentage:",
      P_rain_given_cloudy * 100, "%")
```

Spam email

```python
P_spam = 0.30
P_offer_given_spam = 0.80
P_offer_given_ham = 0.10

P_ham = 1 - P_spam

# Bayes theorem
P_spam_given_offer = (
    P_offer_given_spam * P_spam
) / (
    (P_offer_given_spam * P_spam) +
    (P_offer_given_ham * P_ham)
)

print("Probability that the email is spam:", P_spam_given_offer)
print("Percentage:", P_spam_given_offer * 100, "%")
```
