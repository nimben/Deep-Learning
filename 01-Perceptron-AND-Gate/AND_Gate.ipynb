```python
X = [[0,0],[0,1],[1,0],[1,1]]

y =[0,0,0,1]

w1 =0.0
w2=0.0
b=0.0

lr =0.1

epochs =1000

def activation(value):
  if value > 0:
    return 1
  else:
    return 0

for epoch in range(epochs):
  for i in range(len(X)):
    x1 = X[i][0]
    x2 = X[i][1]

    weighted_sum = x1 * w1 + x2 * w2 + b
    pred = activation(weighted_sum)

    error = y[i]-pred

    if error != 0:
      w1=w1+lr*error*x1
      w2=w2+lr*error*x2

      b=b+lr*error

print("Final weights:")
print("w1=",w1)
print("w2=",w2)
print("bias=",b)


print("\nAND Gate Predictions:")

for inputs in X:

    x1 = inputs[0]
    x2 = inputs[1]

    weighted_sum = w1*x1 + w2*x2 + b
    predicted = activation(weighted_sum)

    print(inputs, "->", predicted)





```
